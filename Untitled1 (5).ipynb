```python

import h5py
import numpy as np
import matplotlib.pyplot as plt

file_path = "/content/" + metadata["name"]

with h5py.File(file_path, "r") as f:
    base = "science/LSAR/GCOV/grids/frequencyA/"

    # Downsample for faster visualization
    hh = f[base + "HHHH"][::20, ::20].astype(np.float32)
    hv = f[base + "HVHV"][::20, ::20].astype(np.float32)

# Convert positive linear power values to dB
def to_db(data):
    result = np.full(data.shape, np.nan, dtype=np.float32)
    valid = np.isfinite(data) & (data > 0)
    result[valid] = 10 * np.log10(data[valid])
    return result

hh_db = to_db(hh)
hv_db = to_db(hv)

# Build a false-colour composite
ratio_db = hh_db - hv_db

def percentile_stretch(data, low=2, high=98):
    valid = np.isfinite(data)
    out = np.zeros(data.shape, dtype=np.float32)

    if valid.any():
        lo, hi = np.nanpercentile(data[valid], [low, high])
        if hi > lo:
            out[valid] = np.clip((data[valid] - lo) / (hi - lo), 0, 1)

    return out

red = percentile_stretch(hh_db)
green = percentile_stretch(hv_db)
blue = percentile_stretch(ratio_db)

rgb = np.stack([red, green, blue], axis=-1)

# Hide pixels invalid in either channel
valid = np.isfinite(hh_db) & np.isfinite(hv_db)
rgb[~valid] = 0

plt.figure(figsize=(12, 8))
plt.imshow(rgb)
plt.title("NISAR False-Colour Composite (HHHH, HVHV)")
plt.axis("off")
plt.tight_layout()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

BASE = "science/LSAR/GCOV/grids/frequencyA/"
STRIDE = 20

with h5py.File(file_path, "r") as f:
    hh_raw = f[BASE + "HHHH"][::STRIDE, ::STRIDE].astype(np.float32)
    hv_raw = f[BASE + "HVHV"][::STRIDE, ::STRIDE].astype(np.float32)

def to_db(a):
    out = np.full(a.shape, np.nan, dtype=np.float32)
    valid = np.isfinite(a) & (a > 0)
    out[valid] = 10 * np.log10(a[valid])
    return out

hh_db = to_db(hh_raw)
hv_db = to_db(hv_raw)

# Initially retain pixels with valid, positive measurements
valid = (
    np.isfinite(hh_db) &
    np.isfinite(hv_db)
)

print("Sampled image shape:", hh_raw.shape)
print("Valid paired pixels:", int(valid.sum()))
print("Invalid paired pixels:", int((~valid).sum()))

fig, axes = plt.subplots(1, 2, figsize=(13, 6))

for ax, data, title in zip(
    axes,
    [hh_db, hv_db],
    ["HHHH backscatter (dB)", "HVHV backscatter (dB)"]
):
    if np.isfinite(data).any():
        lo, hi = np.nanpercentile(data[np.isfinite(data)], [2, 98])
        ax.imshow(data, cmap="Greens", vmin=lo, vmax=hi)
    ax.set_title(title)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python

from scipy.ndimage import uniform_filter

def local_std(data, window=5):
    # Fill missing pixels with the finite-data median for filtering only.
    # The original validity mask is retained separately.
    filled = np.where(np.isfinite(data), data, np.nanmedian(data))
    mean = uniform_filter(filled, size=window)
    mean_sq = uniform_filter(filled ** 2, size=window)
    return np.sqrt(np.maximum(mean_sq - mean ** 2, 0))

ratio_db = hh_db - hv_db
hh_std = local_std(hh_db)
hv_std = local_std(hv_db)

feature_stack = np.stack(
    [hh_db, hv_db, ratio_db, hh_std, hv_std],
    axis=-1
)

X_all = feature_stack[valid]

print("Feature names:")
print([
    "HHHH_dB",
    "HVHV_dB",
    "HHHH_minus_HVHV_dB",
    "Local_HHHH_std",
    "Local_HVHV_std"
])
print("Feature matrix shape:", X_all.shape)
print("Missing values in feature matrix:", np.isnan(X_all).sum())
```

```python

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import MiniBatchKMeans

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_all)

kmeans = MiniBatchKMeans(
    n_clusters=4,
    random_state=42,
    batch_size=4096,
    n_init=5
)

cluster_labels = kmeans.fit_predict(X_scaled)

cluster_map = np.full(valid.shape, -1, dtype=np.int16)
cluster_map[valid] = cluster_labels

plt.figure(figsize=(10, 7))
plt.imshow(cluster_map, cmap="viridis", vmin=-1, vmax=3)
plt.title("NISAR Radar-Signature Clusters")
plt.colorbar(label="Cluster ID")
plt.axis("off")
plt.tight_layout()
plt.show()

print("Cluster counts:")
for cluster_id, count in zip(*np.unique(cluster_labels, return_counts=True)):
    print(f"Cluster {cluster_id}: {count:,} pixels")
```

```python

from sklearn.ensemble import IsolationForest

iso = IsolationForest(
    n_estimators=100,
    contamination=0.02,
    random_state=42,
    n_jobs=-1
)

iso.fit(X_scaled)

# Lower decision_function values indicate more unusual observations.
decision = iso.decision_function(X_scaled)
anomaly_score = -decision

anomaly_map = np.full(valid.shape, np.nan, dtype=np.float32)
anomaly_map[valid] = anomaly_score

# Mark the highest-scoring 2% of valid pixels as candidate anomalies.
threshold = np.percentile(anomaly_score, 98)
anomaly_pixels = valid & (anomaly_map >= threshold)

plt.figure(figsize=(10, 7))
plt.imshow(anomaly_map, cmap="magma")
plt.title("NISAR Candidate Anomaly Score")
plt.colorbar(label="Anomaly score (higher = more unusual)")
plt.axis("off")
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 7))
plt.imshow(hh_db, cmap="gray")
plt.imshow(
    np.ma.masked_where(~anomaly_pixels, anomaly_pixels),
    cmap="autumn",
    alpha=0.85
)
plt.title("Candidate Anomalies Overlaid on HHHH")
plt.axis("off")
plt.tight_layout()
plt.show()

print("Anomaly threshold:", float(threshold))
print("Candidate anomalous pixels:", int(anomaly_pixels.sum()))
print("Percentage of valid pixels flagged:",
      round(100 * anomaly_pixels.sum() / valid.sum(), 2))
```

```python

from scipy.ndimage import label

# 8-neighbour connectivity
structure = np.ones((3, 3), dtype=np.uint8)
region_map, region_count = label(anomaly_pixels, structure=structure)

MIN_REGION_PIXELS = 5
regions = []

for region_id in range(1, region_count + 1):
    region_mask = region_map == region_id
    pixel_count = int(region_mask.sum())

    if pixel_count < MIN_REGION_PIXELS:
        continue

    regions.append({
        "Region_ID": region_id,
        "Pixel_Count": pixel_count,
        "Mean_Anomaly_Score": float(
            np.nanmean(anomaly_map[region_mask])
        ),
        "Mean_HHHH_dB": float(
            np.nanmean(hh_db[region_mask])
        ),
        "Mean_HVHV_dB": float(
            np.nanmean(hv_db[region_mask])
        ),
        "Mean_Cluster_ID": float(
            np.mean(cluster_map[region_mask])
        )
    })

regions_df = pd.DataFrame(regions)

if not regions_df.empty:
    regions_df = regions_df.sort_values(
        "Mean_Anomaly_Score", ascending=False
    ).reset_index(drop=True)

display(regions_df.head(20))
print("Candidate regions retained:", len(regions_df))

plt.figure(figsize=(10, 7))
plt.imshow(hh_db, cmap="gray")
plt.imshow(
    np.ma.masked_where(region_map == 0, region_map),
    cmap="autumn",
    alpha=0.85
)
plt.title("Candidate Anomaly Regions")
plt.axis("off")
plt.tight_layout()
plt.show()
```
